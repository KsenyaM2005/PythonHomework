3. Реализуйте программу «угадай число». Пользователь загадывает число, на входе указывает интервал, в котором находится загаданное число. Программа «угадывает» число до тех пор, пока не выдаст корректный ответ. Допустимые вопросы: Число равно N? Число меньше N? (возможно использовать метод дихотомии)

```python
print('Введите по очереди нижнюю и верхнюю границу интервала')
m = int(input())
M = int(input())
mid = (m+M) // 2
flag = True
while m < M and flag:
    mid = (m+M) // 2
    print('Число равно {}? [y/n]'.format(mid))
    ans = input()
    if ans == 'y':
        print('Ура, я угадал!')
        flag = False
    else:
        print('Число меньше {}? [y/n]'.format(mid))
        ans = input()
        if ans == 'y':
            M = mid
        else:
            m = mid+1
```

4. Пользователь вводит числа, большие 0, критерий окончания ввода – число 0. Вывести наибольшее среди всех чисел.

```python
M = 0
while True:
    x = int(input())
    if x == 0:
        if M == 0:
            print("Не было введено положительных чисел")
        else:
            print(M)
        break
    M = max(x, M)
```

5. Выведите на экран таблицу умножения чисел от одного до девяти.

```python
numbers = range(1,10)
print("\t|\t", end='')
print("\t".join([str(x) for x in numbers]))
print('----|' + '-'* 38)
for left in numbers:
    print("{}\t|\t".format(left), end='')
    print("\t".join([str(left * x) for x in numbers]))
```

2. Задан список из 20 элементов. Каждый элемент – число от 0 до 100. Рассчитать гистограмму значений, квантованную по числу 10: 1й бин – количество элементов списка в диапазоне [0,9], 2й бин – количество элементов списка в диапазоне [10,19]... Вывести результат. Перевести количество в вероятности встречаемости значения из диапазона. Вывести результат.

```python
n = 20
array = [random.randint(0, 100) for _ in range(n)]

print(array)
```

```python
array.sort()
print("Сортированный список:")
print(array)
print()

histogram = [0] * 10
bounds = range(10, 101, 10)
binIdx = 0

for x in array:
    while x >= bounds[binIdx]:
        binIdx = binIdx + 1
    histogram[binIdx] += 1

print("Верхние границы и значения гистограммы:")
print('\t'.join(f"{b:6}" for b in bounds))
print('\t'.join(f"{x:6}" for x in histogram))

print()

print("Вероятности встречаемости чисел из диапазонов:")
print('\t'.join(f"{b:6}" for b in bounds))
print('\t'.join(f"{x/sum(histogram):6.2f}" for x in histogram))
```

3. Заданы два вектора, размерности N. Найти покомпонентные сумму, умножение, вывести результат. Для вектора с бОльшей нормой рассчитать результат умножения на скаляр. Скаляр, размерность задаются из консоли. Заполнение векторов – при помощи генератора псевдослучайных чисел.

```python
def operVector(a, b, operation):
    if len(a) != len(b):
        raise ValueError("Неравные длины векторов")
    ans = [0] * len(a)
    for i in range(len(a)):
        ans[i] = operation(a[i], b[i])
    return ans


def sumNumbers(a, b):
    return a + b

def multNumbers(a, b):
    return a * b


def sumVectors(a, b):
    return operVector(a, b, sumNumbers)

def multVectors(a, b):
    return operVector(a, b, multNumbers)

def multVecAndNum(vec, number):
    return operVector(vec, [number]*len(vec), multNumbers)
```
