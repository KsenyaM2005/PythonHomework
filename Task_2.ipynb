1. Реализовать функцию (или набор функций), которые выводят на экран таблицы умножения, деления, вычитания, сложения для чисел 1..9. Выбор таблицы осуществляется пользователем.

```python
import random


def printTable(numbers, operation):
    print("\t|\t", end='')
    print("\t".join([str(x) for x in numbers]))
    print('----|' + '-'* 38)
    for left in numbers:
        print("{}\t|\t".format(left), end='')
        print("\t".join([str(operation(left, right)) for right in numbers]))
```

```python
def sumNumbers(a, b):
    return a + b

def multNumbers(a, b):
    return a * b

def subNumbers(a, b):
    return a - b

def divNumbers(a, b):
    if b != 0:
        return a / b
    else:
        raise ValueError("Деление на ноль!")
```

```python
def func():
    oper = input("Введите нужную операцию: [+ | - | * | /]").strip()

    numbers = range(1, 10)
    if oper == '+':
        printTable(numbers, sumNumbers)
    elif oper == '-':
        printTable(numbers, subNumbers)
    elif oper == '*':
        printTable(numbers, multNumbers)
    elif oper == '/':
        printTable(numbers, divNumbers)
    else:
        raise ValueError("Непредусмотренная операция!")
```

2. Реализуйте функции:
    a. Создающая вектор заданной длины N, заполнение – случайные числа от 0..1
    b. Создающая матрицу MxN, заполнение – случайные числа 0..1
    c. Умножающая матрицу на вектор (см. предыдущее ДЗ)
    d. Печатающую матрицу
    e. Печатающую вектор
    f. Находящую сумму диагональных элементов матрицы
    g. Реализующая двумерную свертку изображения.

```python
import random
```

```python
# a. Создающая вектор заданной длины N, заполнение – случайные числа от 0..1
def randomVec(N):
    return [random.uniform(0,1) for _ in range(N)]

# b. Создающая матрицу MxN, заполнение – случайные числа 0..1
def randomMat(M, N):
    return [[random.uniform(0,1) for i in range(N)] for j in range(M)]

# c. Умножающая матрицу на вектор
def multMatVec(M, v):
    ans = [0]*len(M)

    for i in range(len(M)):
        row = M[i]
        if len(row) != len(v):
            raise ValueError("Несоответствие размеров!")
        ans[i] = sum([row[j]*v[j] for j in range(len(row))])
    return ans


# e. Печатающую вектор
def printVec(v):
    print('[' + "\t".join([str(x) for x in v]) + ']')


# d. Печатающую матрицу
def printMat(M):
    print('[')
    for row in M:
        printVec(row)
    print(']')


# f. Находящую сумму диагональных элементов матрицы
def sumDiag(M):
    ans = 0
    for i in range(min(len(M), len(M[0]))):
        if len(M[i]) != len(M):
            raise ValueError("Неквадратная матрица!")
        ans += M[i][i]
    return ans

# g. Реализующая двумерную свертку изображения.
def conv(M, filter):
    n = len(filter)
    s = len(M) - n + 1
    ans = [[0] * s for _ in range(s)]

    for i in range(s):
        for j in range(s):
            window = [row[j:j+n] for row in M[i:i+n]]
            ans[i][j] = sum(window[a][b] * filter[a][b]
                            for a in range(n) for b in range(n))
    return ans
```

```python
v = randomVec(4)
printVec(v)
```

```python
M = randomMat(5,4)
printMat(M)
```

```python
printVec(multMatVec(M, v))
```

```python
M = randomMat(5,5)
printMat(M)
print(sumDiag(M))
```

```python
M = randomMat(5,5)
printMat(M)
filter = randomMat(3,3)
printMat(filter)
print()

printMat(conv(M, filter))
```

```python
def decoratorMultiCanal(func):
    def wrapper(*argv):
        M = argv[0]
        result = [0] * len(M)
        for i, canal in enumerate(M):
            result[i] = func(canal, *argv[1:])
        return result
    return wrapper

def decoratorMultiCanal(func):
    def wrapper(*argv):
        M = argv[0]
        result = [0] * len(M)
        for i, canal in enumerate(M):
            result[i] = func(canal, *argv[1:])
        return result
    return wrapper
```

4. Реализовать функцию автоматического конвертирования для цветового вектора из модели RGB в модель YIQ (и обратно). Цвет кодируется вектoром из 4х значений: первые три компоненты- цвет, четвертая – тип (0 – RGB, 1 – YIQ).

```python
def RGBtoYIC(color):
    if color[3] == 0:
        R, G, B = color[0], color[1], color[2]
        Y = 0.299 * R + 0.587 * G + 0.114 * B
        I = 0.596 * R - 0.274 * G - 0.322 * B
        Q = 0.211 * R - 0.522 * G + 0.311 * B
        return [Y, I, Q, 1]
    elif color[3] == 1:
        return color[:3]
    else:
        raise ValueError("Неправильная кодировка типа цветовой модели!")

def decoratorVertorOper(func):
    def wrapper(*argv):
        vec = argv[0]
        result = [0] * len(vec)
        for i, elem in enumerate(vec):
            result[i] = func(elem, *argv[1:])
        return result
    return wrapper

RGBtoYIC_vectorized = decoratorVertorOper(RGBtoYIC)
```

```python
N = 10
vec = [[random.uniform(0,1), random.uniform(0,1), random.uniform(0,1), 0] for _ in range(N)]
printMat(vec)
```

```python
printMat(RGBtoYIC_vectorized(vec))
```
