1. Напишите программу, в которой в бесконечном цикле пользователь вводит два числа, а программа выводит их сумму.

```python
while True:
    a = int(input())
    b = int(input())
    print(a+b)
```

2. Используйте циклы, чтобы вывести на консоль следующий набор символов:

```python
print('* ', end='')
for i in range(1,13):
    if i % 5 == 2:
        print('{} * '.format(i), end='\n')
    elif i % 5 == 0:
        print('{}\n* '.format(i), end='')
    else:
        print('{} * '.format(i), end='')
```

3. Реализуйте программу «угадай число». Пользователь загадывает число, на входе указывает интервал, в котором находится загаданное число. Программа «угадывает» число до тех пор, пока не выдаст корректный ответ. Допустимые вопросы: Число равно N? Число меньше N? (возможно использовать метод дихотомии)

```python
print('Введите по очереди нижнюю и верхнюю границу интервала')
m = int(input())
M = int(input())
mid = (m+M) // 2
flag = True
while m < M and flag:
    mid = (m+M) // 2
    print('Число равно {}? [y/n]'.format(mid))
    ans = input()
    if ans == 'y':
        print('Ура, я угадал!')
        flag = False
    else:
        print('Число меньше {}? [y/n]'.format(mid))
        ans = input()
        if ans == 'y':
            M = mid
        else:
            m = mid+1
```
